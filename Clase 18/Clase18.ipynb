## Agente

Un agente es un programa de software que aprende a tomar decisiones inteligentes. Podemos decir que un agente es un aprendiz en el entorno de RL. Por ejemplo, un jugador de ajedrez puede considerarse un agente ya que aprende a realizar los mejores movimientos (decisiones) para ganar la partida. De manera similar, Mario en un videojuego de Super Mario Bros puede considerarse un agente ya que explora el juego y aprende a realizar los mejores movimientos.



## Ambiente (Entorno)

El entorno es el mundo del agente. El agente permanece dentro del medio ambiente. Por ejemplo, volviendo a nuestro juego de ajedrez, un tablero de ajedrez es un entorno ya que el jugador (agente) aprende a jugar ajedrez dentro del tablero (entorno). De manera similar, en Super Mario Bros, el mundo de Mario es el entorno.

## Estado y acción

El estado suele denotarse por $s$ y es una posición o un momento en el entorno en el que el agente puede estar. Aprendimos que el agente permanece dentro del entorno y que pueden existir muchas posiciones en el entorno en las que el agente puede permanecer, y esas posiciones se llaman estados. Por ejemplo, en nuestro ejemplo del juego de ajedrez, cada posición en el tablero es un estado. 

El agente interactúa con el entorno y pasa de un estado a otro realizando una acción. En el entorno del juego de ajedrez, la acción es el movimiento realizado por el jugador (agente). La acción suele denotarse con $a$.

## Recompensa

Aprendimos que el agente interactúa con un entorno realizando una acción y pasa de un estado a otro. Según la acción, el agente recibe una recompensa. Una recompensa no es más que un valor numérico, por ejemplo: +1 por una buena acción y -1 por una mala acción. 

Sin embargo, ¿cómo decidimos si una acción es buena o mala?

En nuestro ejemplo de juego de ajedrez, si el agente realiza un movimiento en el cual se come una de las piezas del oponente, entonces se considera una buena acción y el agente recibe una recompensa positiva. De manera similar, si el agente hace un movimiento que lleva al oponente a comerse la pieza de ajedrez del agente, entonces se considera una mala acción y el agente recibe una recompensa negativa. La recompensa se denota por $r$.

# La idea básica de RL

Supongamos que estamos enseñando a un perro (agente) a atrapar una pelota. En lugar de enseñarle explícitamente al perro a atrapar una pelota, simplemente lanzamos la pelota y cada vez que el perro la atrapa, le damos una galleta (recompensa). Si el perro no logra atrapar la pelota, no le damos galletas. Entonces, el perro descubrirá qué acción provocó que recibiera una recompensa y repetirá esa acción.

En RL, el agente realiza un proceso de aprendizaje por prueba y error. No se le enseña qué hacer ni cómo hacerlo, solo se le entrega una recompensa positiva cuando realiza una buena acción y una recompensa negativa cuando realiza una mala acción.

El agente comienza realizando una acción aleatoria y, si recibe una recompensa por esa acción, "entiende" que ha actuado bien y repetirá ese comportamiento. Por el contrario, si recibe castigo no volverá a repetirla.

# Procesos de decisión de Markov

- El proceso de decisión de Markov (MDP) proporciona un marco matemático para resolver un problema de RL. 

- Casi todos los problemas de RL se pueden modelar como MDP.

- Lo primero es entender el concepto de propiedad de Markov y cadena de Markov.

## La propiedad de Markov

La propiedad de Markov afirma que el futuro depende sólo del presente y no del pasado.

## La cadena de Markov

La cadena de Markov, también conocida como proceso de Markov, consta de una secuencia de estados que obedecen estrictamente a la propiedad de Markov; es decir, la cadena de Markov es el modelo probabilístico que depende únicamente del estado actual para predecir el siguiente estado y no los estados anteriores, es decir, el futuro es condicionalmente independiente del pasado.

## Proceso de Recompensa de Markov

- El proceso de recompensa de Markov (MRP) es una extensión de la cadena de Markov con una función de recompensa.

- La función de recompensa suele denotarse por $R(s)$ y nos dice la recompensa que obtenemos en cada estado. 

- Por tanto, el MRP consta de estados $s$, una probabilidad de transición $𝑃(s'|s)$ y una función de recompensa $R(s)$.

## Proceso de Decisión de Markov

El Proceso de Decisión de Markov (MDP) permite modelar un problema de RL y es una extensión del MRP con acciones.

# Conceptos fundamentales de RL

## Promedio

Digamos que tenemos una variable X que tiene los valores 1, 2, 3, 4, 5, 6. Para calcular el valor promedio de X, podemos simplemente sumar todos los valores de X y dividirlos por el número total de valores. Por tanto, el promedio de X es (1+2+3+4+5+6)/6 = 3.5.

## Expectation (Esperanza)

Ahora, supongamos que X es una variable aleatoria. Una variable aleatoria toma valores basados en un experimento aleatorio, como tirar un dado, una moneda, etc. Por lo tanto, cada valor tiene asociada una probabilid. Supongamos que estamos lanzando un dado, entonces los resultados posibles (X) son 1, 2, 3, 4, 5, 6 y la probabilidad de que ocurra cada uno de estos resultados es 1/6.

¿Cómo podemos calcular el valor promedio de la variable aleatoria X? Como cada valor tiene una probabilidad de ocurrir, no podemos simplemente tomar el promedio. Entonces, en lugar de eso, calculamos el promedio ponderado, es decir, la suma de los valores de X multiplicada por sus respectivas probabilidades, y esto se llama Esperanza matemática.

$$ E(x) = \sum_{i=1}^{N} x_ip(x_i) = 1(1/6) + 2(1/6) + 3(1/6) +
4(1/6) + 5 (1/6) + 6(1/6) = 3.5 $$

Podemos categorizar los espacios de acción en dos tipos:
* Espacio de acción discreto
* Espacio de acción continuo

### Espacio de acción discreto

Cuando nuestro espacio de acción consta de acciones discretas, se le llama espacio de acción discreto. Por ejemplo, en el mundo cuadrícula, nuestro espacio de acción consta de cuatro acciones discretas *[up, down, left, right]*.

### Espacio de acción continuo

Consta de acciones que son continuas. Por ejemplo, supongamos que estamos entrenando a un agente para que conduzca un automóvil, entonces nuestro espacio de acción consistirá en varias acciones que tienen valores continuos, como por ejemplo el ángulo de giro de la rueda *[36.62°, 54.5°, ...]*.

## Política (Policy)

Una política define el comportamiento del agente en un entorno. La política le dice al agente qué acción realizar en cada estado. Por ejemplo, moverse hacia Abajo en el estado A, moverse hacia la Derecha en el estado D, y así sucesivamente.

Al interactuar con el entorno por primera vez, inicializamos una política aleatoria (indica acciones aleatorias en cada estado). Luego, el agente redefine su política con base en las recompensas y castigos hasta llegar a una política óptima que indique la acción correcta (que trae consigo una recompensa).

### Política determinista

Una política determinista le dice al agente que realice una acción particular en un estado. Por lo tanto, la política determinista asigna al estado una acción particular y a menudo se denota por 𝜇.

Por ejemplo, dado el estado A, la política determinista 𝜇 le dice al agente que realice la acción Abajo. Esto se puede expresar como: 𝜇(𝐴) = Abajo

Por lo tanto, de acuerdo con la política determinista, cada vez que el agente visita el estado A, realiza la misma acción que es moverse hacia Abajo.

### Política estocástica

Con una política estocástica, dado un estado, la política estocástica devolverá una distribución de probabilidad sobre un espacio de acción. Entonces, en lugar de realizar la misma acción cada vez que el agente visita el estado, el agente realiza acciones diferentes cada vez según una distribución de probabilidad devuelta por la política estocástica. Por ejemplo, dado un estado A, la política estocástica devuelve la distribución de probabilidad sobre el espacio de acción como $[0.1, 0.7, 0.1, 0.1]$.

## Tareas en RL

Una tarea en RL se puede clasificar como:
* Tarea episódica: tiene un estado terminal/final. Por ejemplo, en un juego de carreras de auto, partimos del punto de partida (estado inicial) y llegamos a la meta (estado terminal).
* Tarea continua: no contienen episodios y, por lo tanto, no tienen ningún estado terminal. Por ejemplo, un robot asistente personal no tiene estado terminal.

## Horizonte

Horizonte es el paso de tiempo hasta el cual el agente interactúa con el entorno. Podemos clasificar el horizonte en dos categorías:
* Horizonte finito: la interacción agente-entorno se detiene en un lapso de tiempo particular.
* Horizonte infinito: la interacción agente-entorno nunca se detiene.

## Factor de retorno

Un retorno (a veces llamado rendimiento) se denota por $R$ o $G$, y se puede definir como la suma de las recompensas obtenidas por el agente en un episodio.

    <img src="img/episode-example.jpg" alt="" width="80%" align="center" />
</div>

Por ejemplo, el retorno de la trayectoria anterior es $𝑅(𝜏) = 2 + 2 + 1 + 2 = 7$.

## Objetivo

- Podemos decir que el objetivo de nuestro agente es maximizar el retorno, es decir, maximizar la suma de recompensas (recompensas acumulativas) obtenidas a lo largo del episodio. 

- Pero, ¿cómo podemos maximizar el retorno? Podemos maximizar el retorno si realizamos la acción correcta en cada estado. 

- Bien, ¿cómo podemos realizar la acción correcta en cada estado? Podemos realizar la acción correcta en cada estado utilizando la política óptima. Por lo tanto, la política óptima es aquella que le da a nuestro agente el máximo rendimiento (suma de recompensas) al realizar la acción correcta en cada estado.

## La función de valor

La función de valor (también llamada función de valor de estado) denota el valor del estado. El valor de un estado es el retorno que obtendría un agente a partir de ese estado siguiendo la política 𝜋. El valor de un estado o función de valor generalmente se denota por $V(s)$.

### Cálculo del valor

La trayectoria anterior se genera usando la política 𝜋, por lo que podemos decir que el valor de un estado es el retorno (suma de recompensas) de la trayectoria que comienza en ese estado: 

* El valor del estado A es el retorno de la trayectoria que comienza en el estado A. Por tanto, $V(A) = 1+1 -1+1 = 2$.
* El valor del estado D es el retorno de la trayectoria que parte del estado D. Así, $V(D) = 1-1+1= 1$.
* El valor del estado E es el retorno de la trayectoria que parte del estado E. Por tanto, $V(E) = -1+1 = 0$.
* El valor del estado H es el retorno de la trayectoria que parte del estado H. Por tanto, $V(H) = 1$.

# Q function

Una función Q, también llamada función de valor de estado-acción, denota el valor de un par de estado-acción. El valor de un par estado-acción es el rendimiento que obtendría el agente al partir del estado $s$ y realizando la acción según la política 𝜋. El valor de un par estado-acción o función Q generalmente se denota por $Q(s,a)$.

Tenga en cuenta que la única diferencia entre la función de valor y la función Q es que en la función de valor calculamos el valor de un estado, mientras que en la función Q calculamos el valor de un par estado-acción.

# Model-based and model-free learning

* Aprendizaje basado en modelos: el agente tendrá una descripción completa del entorno, es decir, conoce la probabilidad de transición de su entorno. Así, en el aprendizaje basado en modelos, el agente utiliza la dinámica del modelo para encontrar la política óptima.

* Aprendizaje sin modelos: se produce cuando el agente no conoce la dinámica del modelo de su entorno, es decir, intenta encontrar la política óptima sin la dinámica del modelo.

# Diferentes tipos de ambientes

Ya aprendimos que el medio ambiente es el mundo del agente y el agente vive/permanece dentro del medio ambiente. Podemos categorizar el entorno en diferentes tipos.

## Entorno determinista

En un entorno determinista, estamos seguros de que cuando un agente realiza la acción $a$ en el estado $s$, siempre alcanza el estado $𝑠′$. Por ejemplo, supongamos que el agente está en el estado A. Si se mueve hacia Abajo, siempre llegará al estado D.

## Entorno estocástico

En un entorno estocástico, no podemos decir que al realizar la acción $a$ en el estado $s$ el agente siempre alcanza el estado $s'$ porque habrá cierta aleatoriedad asociada con el entorno. Por ejemplo, supongamos que nuestro agente está en el estado A; ahora, si se mueve hacia Abajo llega al estado D el 70% de las veces y al estado B el 30% de las veces.

## Entornos discretos y continuos

* Entorno discreto: aquel en el que el espacio de acción del entorno es discreto.
* Entorno continuo: aquel en el que el el espacio de acción del entorno es continuo.

## Entornos episódico y no-episódico

* Entorno episódico: la acción actual de un agente no afectará las acciones futuras y, por lo tanto, un entorno episódico también se denomina entorno no secuencial.
* Entorno no episódico: la acción actual de un agente afectará las acciones futuras y, por lo tanto, un entorno no episódico también se denomina entorno secuencial. Por ejemplo, un tablero de ajedrez es un entorno secuencial ya que la acción actual del agente afectará las acciones futuras en una partida de ajedrez.

## Entornos de agente único y multiagente

* Entorno de agente único: Cuando nuestro entorno consta de un solo agente.
* Entorno multiagente: Cuando nuestro entorno consta de múltiples agentes.

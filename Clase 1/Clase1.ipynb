# Entrenamiento

Un modelo recibe y aprende a partir de un conjunto de entrenamiento (*training set*) que contiene ejemplos $(\mathbf{x}_i,y_i)$, donde $\mathbf{x_i} = (x_{i,1},\dots,x_{i,d})$ es la entrada observada con $d$ características (*features*) discretas y/o continuas para la salida $y_i$, que puede ser discreta (clasificación) o continua (regresión) dependiendo del problema.

# Testing

La fase de testing permite evaluar si el modelo produce la salida correcta $y_t$ para ejemplos futuros $\mathbf{x_t}=x_{t,1},\dots,x_{t,d}$.

# EL SOBREAJUSTE TIENE MUCHAS CARAS

Todo el mundo en el aprendizaje automático sabe sobre el sobreajuste, pero se presenta en muchas formas que no son inmediatamente obvias.

## Falsa sensación de éxito

Si su modelo tiene una precisión del 100% sobre los datos de entrenamiento, pero solo un 50% sobre los datos de testing, cuando de hecho podría haber logrado un 75% de precisión en ambos datos, simplemente está codificando peculiaridades aleatorias en los datos y por lo tanto no está basado en la realidad. Este problema se denomina sobreajuste (*overfitting*) y es la pesadilla del aprendizaje automático.

<div align="center">
    <img src="table.png" alt="" width="40%" align="center" />
</div>

## Sesgo y varianza
Una forma de entenderlo es descomponiendo el error en sesgo y varianza:

- Sesgo: Tendencia a aprender constantemente la misma cosa errónea.
- Varianza: Tendencia a aprender cosas aleatorias independientes del objetivo real.

# LA INTUICIÓN FALLA EN ALTAS DIMENSIONES

- Después del overfitting, el mayor problema en machine learning es la **maldición de la dimensionalidad**.

- Muchos algoritmos que funcionan bien en pocas dimensiones, se vuelven intratables cuando el input es de alta-dimensionalidad.

- Generalizar correctamente se vuelve exponencialmente más difícil cuando la dimensionalidad aumenta porque el conjunto de entrenamiento cubre cada vez menos el espacio de entrada.

## Los datos se vuelven escasos y dispersos


<div align="center">
    <img src="sparseness.png" alt="" width="100%" align="center" />
</div>

Si tenemos 50 datos y dividimos el espacio en 5 partes, en 1D quedan 50/5 = 10 datos por parte, en 2D quedan 50/(5x5) = 2 datos por parte, y en 3D quedan 50/(5x5x5) = 0.4 datos por parte.

## Nuestro mundo es en 3D

Incluso nuestras intuiciones, que provienen de un mundo tridimensional, no aplican en altas dimensiones. Por ejemplo:

- La mayor parte del volumen de una naranja de altas dimensiones está en la cáscara y no en la pulpa.

## La "Bendición de la no uniformidad"

- Afortunadamente, hay un efecto que contrarresta en parte esta maldición.

- Los datos no están distribuidos uniformemente por todo el espacio.

- Por lo tanto, podemos aplicar técnicas de reducción de la dimensionalidad.

# LAS GARANTÍAS TEÓRICAS NO SON LO QUE PARECEN

Los papers de machine learning están llenos de garantías teóricas. El tipo más común es un límite en el número de ejemplos necesarios para asegurar una buena generalización o dados datos infinitos, se garantiza que el modelo encontrará la respuesta correcta.

## Cuidado con las garantías
Desafortunadamente, las garantías de este tipo deben tomarse con cautela.

- El rol de las garantías teóricas no es servir como criterio para tomar decisiones prácticas, sino como fuente de entendimiento para el diseño de algoritmos.

- Muy útiles por la estrecha relación entre la teoría y la práctica. Pero, el aprendizaje es un fenómeno complejo, y el hecho de que un modelo tenga una justificación teórica y funcione en la práctica no significa que lo primero sea la razón de lo segundo.

# FEATURE ENGINEERING ES CLAVE

Un factor clave en el éxito o fracaso de una predicción son las *features* utilizadas.

<div align="center">
    <img src="features-in-ML.png" alt="" width="70%" align="center" />
</div>

Si se tienen muchas características independientes que se relacionan bien con la clase, el aprendizaje es sencillo.

Por otro lado, si la clase es una función muy compleja de las *features*, es posible que se no pueda aprender.

## La clave del éxito
A menudo, los datos sin procesar no están en una forma que sea adecuada para el aprendizaje, pero se puede construir características a partir de ellos que lo sean. Por lo general, aquí es donde se concentra la mayor parte del esfuerzo en un proyecto de aprendizaje automático. A menudo es también una de las partes más interesantes, donde la intuición, la creatividad y la "magia negra" son tan importantes como la parte técnica.

Los novatos a menudo se sorprenden por el poco tiempo que se dedica en realidad al aprendizaje automático en un proyecto de aprendizaje automático. Pero tiene sentido si considera el tiempo que lleva recopilar datos, integrarlos, limpiarlos y preprocesarlos, y cuánto ensayo y error puede implicar el diseño de características.

## Es un proceso iterativo

Además, el aprendizaje automático no es un proceso lineal, sino un proceso iterativo que consiste en ejecutar el modelo, analizar los resultados, modificar los datos, ajustar el modelo, y repetir esto muchas veces hasta lograr el éxito.

El aprendizaje suele ser la parte más rápida de esto y la ingeniería de características la más difícil porque es específica del dominio, mientras que los modelos pueden ser en gran medida de propósito general. Sin embargo, no existe una frontera clara entre los dos.

# MÁS DATOS PUEDEN VENCER A LA INTELIGENCIA

Suponiendo que se tiene el mejor conjunto de features posible, pero no se logra la precisión necesaria. ¿Qué hacer?

Hay dos alternativas posibles:

- Diseñar un mejor algoritmo de aprendizaje. 
- Reunir más datos.

Los investigadores de aprendizaje automático se preocupan principalmente por lo primero, pero pragmáticamente, el camino más rápido hacia el éxito suele ser obtener más datos.

## Regla no escrita

Un algoritmo débil con muchos datos supera a uno más complejo con pocos datos (después de todo, el aprendizaje automático se trata de dejar que los datos hagan el trabajo pesado).

# SIMPLICIDAD NO IMPLICA PRECISIÓN

## Principio "La navaja de Occam"

“En igualdad de condiciones, la explicación más sencilla suele ser la correcta, no deben multiplicarse las entidades sin necesidad”

<div align="center">
    <img src="occam.png" alt="" width="30%" align="center" />
</div>

## La simplicidad por ser simple

- En machine learning esto se puede entender así: "Entre dos modelos con el mismo error en training, el más simple tendrá el menor error en testing."

- Sin embargo, se pueden encontrar muchos ejemplos que afirman lo contrario.

La conclusión sobre este principio es que una hipótesis simple es preferible porque la simplicidad es una virtud en sí misma, pero no implica un aumento en la precisión.

# REPRESENTABLE NO IMPLICA APRENDIBLE

Sólo porque una función pueda ser representada no significa que pueda ser aprendida.

La pregunta clave no es ¿Puede representarse? es **¿Puede aprenderse?**

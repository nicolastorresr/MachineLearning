# El método Monte Carlo

El método Monte Carlo es una técnica estadística que se utiliza para encontrar una solución aproximada mediante muestreo.

La aproximación será mejor cuando el tamaño de la muestra sea grande.

# Monte Carlo prediction

Para aproximar el valor del estado usando el método de Monte Carlo, tomamos muestras de episodios (trayectorias) siguiendo la política dada 𝜋 durante algunas N veces y luego calculamos el valor del estado como el rendimiento promedio de un estado en todos los episodios muestreados.

$$ V(s) \approx \frac{1}{N} \sum_{i=1}^N R_i(s)  $$

### Calcular Valor Estado A

Por lo tanto, generamos tres episodios y calculamos el retorno del estado A en los tres episodios. Ahora bien, ¿cómo podemos calcular el valor del estado A? Si en el método de Monte Carlo, el valor de un estado se puede aproximar calculando el rendimiento promedio del estado en algunos episodios (trayectorias):

$$ V(s) \approx \frac{1}{N} \sum_{i=1}^N R_i(s)  $$


Necesitamos calcular el valor del estado A, por lo que podemos calcularlo simplemente tomando el rendimiento promedio del estado A en los 3 episodios como:

$$ V(A) \approx \frac{1}{3} \sum_{i=1}^3 R_i(A) = \frac{1}{3} (R_1(A) + R_2(A) + R_3(A)) = \frac{4+2+4}{3} = 3.3 $$

### Iteración 1:

La etapa 2, genera un episodio usando la política de entrada dada 𝜋:

<div align="center">
    <img src="img/step2.jpg" alt="" width="60%" align="center" />
</div>

$$ rewards = [1, 1] $$


Primero, calculamos el retorno del estado $S_0$ (suma de recompensas de $S_0$):

\begin{equation} 
\begin{split}
R(S_0) & = & sum(rewards[0: ]) \\
 & = & sum([1, 1]) \\
 & = & 2 \\
\end{split}
\end{equation}

Actualice el retorno total del estado $S_0$ en nuestra tabla como:

\begin{equation} 
\begin{split}
total\_returns(S_0) & = & total\_returns(S_0) + R(S_0) \\
 & = & 0 + 2 \\
 & = & 2 \\
\end{split}
\end{equation}

Actualice la cantidad de veces que se visita el estado $S_0$ en nuestra tabla como:

\begin{equation} 
\begin{split}
N(S_0) & = & N(S_0) + 1 \\
 & = & 0 + 1 \\
 & = & 1 \\
\end{split}
\end{equation}


Ahora, calculemos el retorno del estado $S_1$ (suma de recompensas de $S_1$):

\begin{equation} 
\begin{split}
R(S_1) & = & sum(rewards[1: ]) \\
 & = & sum([1]) \\
 & = & 1 \\
\end{split}
\end{equation}

Actualice el retorno total del estado $S_1$ en nuestra tabla como:

\begin{equation} 
\begin{split}
total\_returns(S_1) & = & total\_returns(S_1) + R(S_1) \\
 & = & 0 + 1 \\
 & = & 1 \\
\end{split}
\end{equation}

Actualice la cantidad de veces que se visita el estado $S_1$ en nuestra tabla como:

\begin{equation} 
\begin{split}
N(S_1) & = & N(S_1) + 1 \\
 & = & 0 + 1 \\
 & = & 1 \\
\end{split}
\end{equation}

### The blackjack environment in the Gym library

```python
import gym
import pandas as pd
from collections import defaultdict
```

```python
env = gym.make('Blackjack-v1')
```

#### Estado

En el entorno del blackjack, el estado se representa como una tupla que consta de tres valores:
1. El valor de la suma de nuestras cartas.
2. El valor nominal de una de las cartas del crupier.
3. Valor booleano: Verdadero si tenemos un As utilizable y Falso si no tenemos un As utilizable.

```python
state = env.reset()
print(state)
```

#### Definiendo una política

Definimos la política de entrada cuya función de valor se predecirá en los próximos pasos.

Nuestra función de política recibe el estado como entrada y si la suma de nuestras cartas (índice `0` del estado) es mayor que 19, entonces devolverá la acción 0 (no pedir otra carta), de lo contrario devolverá la acción 1 (pedir otra carta).

```python
def policy(state):
    return 0 if state[0] > 19 else 1
```

```python
print(policy(state))
```

#### Generando un episodio

A continuación, generamos un episodio usando la política dada, por lo que definimos una función llamada `generate_episode`, que toma la política como entrada y genera el episodio usando la política dada.

```python
num_timestep = 100
```

```python
def generate_episode(policy):

    #let's define a list called episode for storing the episode
    episode = []

    #initialize the state by resetting the environment
    state = env.reset()

    #then for each time step
    for i in range(num_timestep):

        #select the action according to the given policy
        action = policy(state)

        #perform the action and store the next state information
        next_state, reward, done, info = env.step(action)

        #store the state, action, reward into our episode list
        episode.append((state, action, reward))

        #If the next state is a final state then break the loop else update the next state to the current state
        if done:
            break

        state = next_state

    return episode
```

```python
generate_episode(policy)
```

La salida tiene la forma de `[(estado, acción, recompensa)]` y como se dijo anteriormente, `estado` es una tupla con 3 elementos. Entonces, por ejemplo, en el estado `((13, 1, True), 1, 0.0)`, realizamos la acción `1` (pedir una carta) y recibimos una recompensa de `0`. La recompensa se asigna de la siguiente manera: `+1.0` si ganamos el juego, `-1.0` si perdemos el juego, `0` si el juego es un empate.

#### Calcular la función de valor

Aprendimos que para predecir la función de valor, generamos varios episodios utilizando la política dada y calculamos el valor del estado como un rendimiento promedio en varios episodios. Veamos cómo implementar eso.

```python
total_return = defaultdict(float)
N = defaultdict(int)
```

```python
num_iterations = 500000
```

```python
#then, for every iteration
for i in range(num_iterations):

    #generate the episode using the given policy, that is, generate an episode using the policy
    #function we defined earlier
    episode = generate_episode(policy)

    #store all the states, actions, rewards obtained from the episode
    states, actions, rewards = zip(*episode)

    #then for each step in the episode
    for t, state in enumerate(states):

            #compute the return R of the state as the sum of reward
            R = (sum(rewards[t:]))

            #update the total_return of the state
            total_return[state] =  total_return[state] + R

            #update the number of times the state is visited in the episode
            N[state] =  N[state] + 1
```

```python
total_return = pd.DataFrame(total_return.items(),columns=['state', 'total_return'])
```

```python
N = pd.DataFrame(N.items(),columns=['state', 'N'])
```

```python
df = pd.merge(total_return, N, on="state")
```

```python
df.head(10)
```

```python
df['value'] = df['total_return']/df['N']
```

```python
df.head(10)
```

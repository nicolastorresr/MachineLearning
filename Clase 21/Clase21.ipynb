# Model-free vs Model-based

Los enfoques de RL *Model-free*, que aprenden a predecir acciones exitosas mediante ensayo y error, han permitido que los agentes jueguen juegos de Atari y que derroten a campeones mundiales en Starcraft II, pero requieren grandes cantidades de interacción con el entorno, lo que limita su utilidad para escenarios del mundo real.

Por el contrario, los enfoques de RL *Model-based* aprenden un modelo simplificado del entorno. Este modelo del mundo posibilita al agente predecir los resultados de posibles secuencias de acciones, lo que le permite jugar con escenarios hipotéticos para tomar decisiones informadas en nuevas situaciones, reduciendo así el ensayo y error necesario para alcanzar los objetivos.

## SketchRNN

SketchRNN es un modelo generativo basado en redes neuronales recurrentes (RNNs) diseñado para crear y completar dibujos de líneas simples, como bocetos de figuras o formas.

# Dreamer

Dreamer es un modelo de Google DeepMind, y al igual que en World Models utiliza:

- **Un modelo latente:** Dreamer crea un espacio latente que codifica las dinámicas del entorno, lo que permite predicciones precisas de las consecuencias de acciones futuras.
- **Simulación interna:** Los "episodios imaginados" o simulaciones internas le permiten al agente anticipar resultados antes de actuar en el entorno real.
- **Planificación en el espacio latente:** Ambos modelos aprovechan las predicciones en el espacio latente para optimizar la política del agente, lo que ahorra recursos y mejora la eficiencia.

Dreamer aprende un modelo del mundo a partir de su experiencia. Utilizando imágenes pasadas ($o_1–o_3$) y acciones ($a_1–a_2$), calcula una secuencia de estados compactos del modelo (círculos verdes) a partir de los cuales reconstruye las imágenes ($ô_1–ô_3$) y predice las recompensas ($r̂_1–r̂_3$).

# Learning the World Model

La predicción anticipada mediante estados de modelos compactos permite realizar predicciones a largo plazo en entornos complejos. Aquí se muestran dos secuencias que el agente no ha encontrado antes. Dadas cinco imágenes de entrada, el modelo las reconstruye y predice las imágenes futuras hasta el paso de tiempo 50.

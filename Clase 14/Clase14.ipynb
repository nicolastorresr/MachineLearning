# ¿Qué es el modelado generativo?

Un modelo generativo se puede definir ampliamente de la siguiente manera:

<blockquote>
Un modelo generativo describe cómo se genera un conjunto de datos, en términos de un modelo probabilístico. Al tomar muestras de este modelo, podemos generar nuevos datos.
</blockquote>

# Desafío de modelado generativo

Comencemos con un desafío de modelado generativo en sólo dos dimensiones. Una cierta regla, $p_{data}$, se utilizó para generar el siguiente conjunto de puntos:

<div align="center">
    <img src="img/p_data.png" alt="" width="50%" align="center" />
</div>

El desafío es determinar un nuevo punto $𝐱 = (x_1, x_2)$ en el espacio que haya sido generado por la misma regla.

## Nuestro primer modelo generativo

Sea un modelo $p_{model}$ que forma un cuadro rectangular, dentro del cual se pueden encontrar puntos y un área fuera del cuadro donde no hay posibilidad de encontrar ningún punto. Para generar un nuevo punto $𝐱 = (x_1, x_2)$, simplemente puede elegir un punto al azar dentro del cuadro o, más formalmente, una **muestra** de la distribución del modelo $p_{model}$.

<div align="center">
    <img src="img/p_model.png" alt="" width="50%" align="center" />
</div>

## El marco de modelado generativo

- Tenemos un conjunto de datos de observaciones $𝐗$. 
- Suponemos que las observaciones se han generado según alguna distribución desconocida, $p_{data}$. 
- Queremos construir un modelo generativo $p_{model}$ que imite $p_{data}$. Si logramos este objetivo, podemos tomar muestras de $p_{model}$ para generar observaciones que parecen haber sido extraídas de $p_{data}$. 

### Consideraciones

- $p_{model}$ debe generar ejemplos que parecen haber sido extraídos de $p_{data}$. 
- $p_{model}$ debe generar ejemplos que sean adecuadamente diferentes de las observaciones en $𝐗$. En otras palabras, el modelo no debería simplemente reproducir exactamente las cosas que ya ha visto.

## Transformación del Manifold

El proceso busca codificar el conjunto de datos de entrenamiento en un espacio latente para que podamos tomar muestras y luego decodificar la muestra al espacio original para generar una nueva observación. Matemáticamente hablando, se busca transformar el altamente no lineal Manifold en el cual se encuentran los datos (por ejemplo, en el espacio de píxeles) en un espacio latente más simple del que se pueda muestrear.

# Probabilidades

Un modelo generativo debe ser probabilístico y no determinista. Si nuestro modelo es simplemente un cálculo fijo, como tomar el valor promedio de cada píxel en el conjunto de datos, no es generativo porque el modelo produce el mismo resultado cada vez. El modelo debe incluir un elemento estocástico (aleatorio) que influya en las muestras individuales generadas por el modelo.

<div align="center">
    <img src="img/dados.png" alt="" width="10%" align="center" />
</div>


Como primer paso, definiremos cuatro términos clave: espacio muestral, función de densidad, modelado paramétrico y estimación de máxima verosimilitud.

## Sample Space (Espacio Muestral)

- El espacio muestral es el conjunto completo de todos los valores que puede tomar una observación $𝐱$. 

- En nuestro desafío anterior, el espacio muestral consta de todos los puntos de latitud y longitud $𝐱 = (x_1, x_2)$ en el mapa mundial. 

- Por ejemplo, $𝐱 = (40.7306, –73.9352)$ es un punto en el espacio muestral (Ciudad de Nueva York) que pertenece a la verdadera distribución generadora de datos. En cambio $𝐱 = (11.3493, 142.1996)$ es un punto en el espacio muestral que no pertenece a la verdadera distribución generadora de datos (está en el mar).

## Probability Density Function (Función De Densidad De Probabilidad)

- Una función de densidad de probabilidad (o simplemente función de densidad), $p(𝐱)$, es una función que mapea un punto $𝐱$ en el espacio muestral a un número entre 0 y 1. 

- La sumatoria de la función de densidad sobre todos los puntos en el espacio muestral debe ser igual a 1, para que sea una distribución de probabilidad bien definida.

- En el ejemplo del mapa mundial, la función de densidad de nuestro modelo es 0 fuera del cuadro naranja y constante dentro del cuadro, de modo que la integral de la función de densidad en todo el espacio muestral es igual a 1.

- Si bien sólo existe una función de densidad verdadera $p_{data}(𝐱)$ que se supone que ha generado el conjunto de datos observables, hay infinitas funciones de densidad que podemos usar para estimar.

## Parametric Modeling (Modelado Paramétrico)

- Un modelo paramétrico, $p_{\theta}(𝐱)$, es una familia de funciones de densidad que se pueden describir utilizando un número finito de parámetros, θ.

- La familia de todos los cuadros posibles que podría dibujar para modelar los puntos anteriores es un ejemplo de modelo paramétrico. En este caso, hay cuatro parámetros: las coordenadas de cada esquina del cuadro.

- Por lo tanto, cada función de densidad $p_{\theta}(𝐱)$ en este modelo paramétrico (es decir, cada cuadro) se puede representar de forma única mediante cuatro números, $\theta = (\theta_1,\theta_2,\theta_3,\theta_4)$.

## Likelihood (Verosimilitud)

- La verosimilitud $\mathcal{L}(\theta | 𝐱)$ de un conjunto de parámetros θ es una función que mide la plausibilidad de θ, dado algún punto observado $𝐱$. 

- Por lo tanto, el modelado paramétrico busca encontrar el valor óptimo $\hat{\theta}$ del conjunto de parámetros que maximiza la probabilidad de observar el conjunto de datos $𝐗$. Esta técnica se llama muy apropiadamente **estimación de máxima verosimilitud**.

<div class="alert alert-block alert-info">
Los modelos generativos pueden considerarse como una forma de estimación de máxima verosimilitud, donde los parámetros $\theta$ son los pesos de las redes neuronales contenidas en el modelo.
</div>
